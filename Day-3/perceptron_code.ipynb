#### Now We will divide our data into two parts input column and output column 

```python
X = df.iloc[:,0:2]  #input column
y = df.iloc[:,-1]  # output column
```

```python
X.sample(10)
```

```python
y.sample(10)
```

*Now we will import our perceptron from sklearn*

```python
pip show scikit-learn
```

```python
pip install scikit-learn
```

```python
from sklearn.linear_model import Perceptron
p = Perceptron()
```

```python
p.fit(X,y)
```

```python
p.coef_
```

```python
p.intercept_
```

```python
pip install mlxtend
```

```python
from mlxtend.plotting import plot_decision_regions
```

```python
plot_decision_regions(X.values,y.values,clf=p,legend=4)
```
